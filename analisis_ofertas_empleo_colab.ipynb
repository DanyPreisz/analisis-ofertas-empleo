# Análisis de Ofertas de Empleo

Proyecto de scraping y análisis listo para **Google Colab**.

**Pregunta:** ¿cómo analizar ofertas de trabajo (habilidades más demandadas)?

**Método:** Scraping + análisis de texto

**Stack:** requests, beautifulsoup4, pandas, matplotlib

Datos de ejemplo.

## 0. Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
print("OK")
```

## 1. Datos de ejemplo (simulados)

```python
ofertas = pd.DataFrame({
    "puesto": ["Data Scientist", "Data Analyst", "ML Engineer", "Data Scientist", "Backend Developer"],
    "habilidades": [
        "python sql machine learning",
        "sql excel tableau",
        "python tensorflow pytorch",
        "python pandas sklearn",
        "python django sql"
    ]
})
print(ofertas)
```

## 2. Extraer habilidades

```python
todas = []
for h in ofertas["habilidades"]:
    todas.extend(h.split())

conteo = Counter(todas)
print(conteo.most_common(10))
```

## 3. Visualización

```python
top = conteo.most_common(8)
skills, counts = zip(*top)

plt.figure(figsize=(8, 4))
plt.barh(skills, counts)
plt.title("Habilidades más demandadas")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## 4. Conclusiones

- Python y SQL suelen ser las habilidades más demandadas.
- Se puede extender a scraping real de LinkedIn, Indeed o Computrabajo.
- Útil para orientar el aprendizaje y el CV.
